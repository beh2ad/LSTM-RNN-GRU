One of the applications of Recurrent neural networks (RNNs) is prediction. In this project, the goal is to use Recurrent neural networks to predict air pollution.
The dataset named "Beijing Pollution Dataset" contains weather information for the city of Beijing. Each row includes the weather data for one hour. The data corresponds to the time period from 2010 to 2015, and the rows are recorded sequentially for every hour from 2010 to 2015.
The data set is accessible through the following link: https://archive.ics.uci.edu/dataset/381/beijing+pm2+5+data

The features in the dataset are:
pollution, dew, temp, pressure, wind_spd, snow, rain

Design a neural network that receives the data as a time series and predicts the air pollution level for the next hour.
(Choose a time window size of the previous 24 hours.)

For each network you design, plot the train and test curves, as well as the true vs. predicted values.
(Use the first 10,000 records as training and validation data, and the remaining data as test data.)

Design the network using GRU, LSTM, and RNN, and compare their run-time and accuracy.
(Measure the training time for a fixed number of epochs.) Analyze the differences.

Investigate the network’s performance using different loss functions and different optimization methods
(e.g., Gradient Descent, RMSProp, Adam).

Time series can have different structures. They may consist of recent consecutive hours, or the same hour across consecutive days, or the same hour across consecutive weeks.
For example:
The same hour over the past 7 days (daily intervals).
The same hour and same weekday over the past 8 weeks (weekly intervals).
Evaluate the network’s performance on daily and weekly time series as well.

Examine the effect of adding a dropout layer to one of the designed networks.
